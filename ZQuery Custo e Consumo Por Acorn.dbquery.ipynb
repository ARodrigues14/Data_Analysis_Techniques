```sql
CREATE OR REPLACE VIEW smart_meters.gold.vw_kpi_demographics AS
SELECT 
    h.acorn_group,
    h.lclid,
    t.type AS tariff_type,
    SUM(f.energy_spent) AS total_energy_kwh,
    SUM(f.money_spent) AS total_revenue_gbp,
    AVG(f.energy_spent) AS avg_energy_per_household 
FROM smart_meters.gold.fact_consumption f
JOIN smart_meters.gold.dim_household h ON f.dim_household_lclid = h.lclid
JOIN smart_meters.gold.dim_tariffs t ON f.dim_tariff_sk = t.sk
GROUP BY 
    h.acorn_group,
    h.lclid,
    t.type;

select * from smart_meters.gold.vw_kpi_demographics
```
