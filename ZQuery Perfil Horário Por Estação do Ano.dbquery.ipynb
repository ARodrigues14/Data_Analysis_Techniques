```sql
CREATE OR REPLACE VIEW smart_meters.gold.vw_load_curve_seasonal AS
SELECT 
    t.season,
    h.lclid,
    t.hour,
    AVG(f.energy_spent) AS avg_hourly_energy_kwh 
FROM smart_meters.gold.fact_consumption f
JOIN smart_meters.gold.dim_time t ON f.dim_time_id = t.id
JOIN smart_meters.gold.dim_household h on f.dim_household_lclid = h.lclid
GROUP BY 
    t.season, 
    h.lclid,
    t.hour
ORDER BY 
    t.season,
    h.lclid,
    t.hour;

SELECT * FROM smart_meters.gold.vw_load_curve_seasonal LIMIT 10
```
