```sql
CREATE OR REPLACE VIEW smart_meters.gold.vw_weather_impact AS
SELECT 
    t.year,
    t.month,
    t.day,
    t.holiday_type,
    w.temp_bin_10,
    w.summary AS weather_summary,
    SUM(f.energy_spent) AS total_energy
    
FROM smart_meters.gold.fact_consumption f
    JOIN smart_meters.gold.dim_time t ON f.dim_time_id = t.id
    JOIN smart_meters.gold.dim_weather w ON f.sk_weather = w.sk_weather

GROUP BY 
    t.year,
    t.month,
    t.day,
    t.holiday_type,
    w.temp_bin_10,
    w.summary
ORDER BY 
    t.year,
    t.month,
    t.day;

SELECT * FROM smart_meters.gold.vw_weather_impact LIMIT 10
```
